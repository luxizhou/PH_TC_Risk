```python
import xarray as xr
import os

def combine_nc_files(file1_path, file2_path, output_path, merge_dim='time'):
    """
    Combine two NetCDF files along a specified dimension and save the result.

    Parameters:
    file1_path (str): Path to the first NetCDF file.
    file2_path (str): Path to the second NetCDF file.
    output_path (str): Path to save the combined NetCDF file.
    merge_dim (str): Dimension along which to combine the files (default: 'time').

    Returns:
    bool: True if successful, False if an error occurs.
    """
    try:
        # Check if input files exist
        if not os.path.exists(file1_path):
            print(f"Error: File {file1_path} does not exist.")
            return False
        if not os.path.exists(file2_path):
            print(f"Error: File {file2_path} does not exist.")
            return False

        # Load the NetCDF files
        ds1 = xr.open_dataset(file1_path)
        ds2 = xr.open_dataset(file2_path)

        # Combine the datasets along the specified dimension
        combined_ds = xr.concat([ds1, ds2], dim=merge_dim)

        # Save the combined dataset to a new NetCDF file
        combined_ds.to_netcdf(output_path)
        print(f"Successfully combined files into {output_path}")

        # Close the datasets to free memory
        ds1.close()
        ds2.close()
        combined_ds.close()

        return True

    except Exception as e:
        print(f"Error combining NetCDF files: {str(e)}")
        return False
```

```python
# Function to check if a point is within the buffer
def is_within_buffer(lon, lat, geometry):
    point = Point(lon, lat)
    return geometry.contains(point)
```

```python
# Function to get TC entry and exit times and name
def get_entry_exit_times(tc_sid, gdf_points, buffer_geometry):
    tc_points = gdf_points[gdf_points['SID'] == tc_sid]
    if tc_points.empty:
        return None, None, None
    tc_points['in_buffer'] = tc_points.geometry.apply(lambda x: buffer_geometry.contains(x))
    buffer_points = tc_points[tc_points['in_buffer']]
    if buffer_points.empty:
        return None, None, None
    entry_time = buffer_points['ISO_TIME'].min()
    exit_time = buffer_points['ISO_TIME'].max()
    tc_name = buffer_points['NAME'].iloc[0] if 'NAME' in buffer_points.columns and not pd.isna(buffer_points['NAME'].iloc[0]) else 'Unknown'
    return entry_time, exit_time, tc_name
```

```python
# Function to plot wind footprint
def plot_wind_footprint(wind_speed_max, tc_sid, entry_time, exit_time, gdf_land, gdf_lines):
    fig = plt.figure(figsize=(12, 8))
    ax = plt.axes(projection=ccrs.PlateCarree())
    ax.set_extent([115, 130, 5, 20], crs=ccrs.PlateCarree())  # Philippines region
    ax.add_feature(cfeature.COASTLINE)
    ax.add_feature(cfeature.BORDERS, linestyle=':')

    # Plot maximum wind speed
    wind_plot = wind_speed_max.plot(ax=ax, cmap=cmap, norm=norm, transform=ccrs.PlateCarree(),
                                    cbar_kwargs={'label': 'Max Wind Speed (m/s)', 'ticks': bins})

    # Overlay Philippines boundary
    gdf_land.plot(ax=ax, facecolor='none', edgecolor='black', linewidth=1)

    # Overlay TC track line
    tc_lines = gdf_lines[gdf_lines['SID'] == tc_sid]
    if not tc_lines.empty:
        tc_lines.plot(ax=ax, color='blue', linewidth=1, label=f'{tc_sid} Track Line')

    ax.gridlines(draw_labels=True, linestyle="--", alpha=0.5)
    ax.set_title(f"Max Wind Footprint of {tc_sid} from {entry_time} to {exit_time}")
    ax.legend()

    # Save plot
    output_file = os.path.join(output_dir, f"wind_footprint_{tc_sid}.png")
    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.close()
    print(f"Saved wind footprint for {tc_sid} to {output_file}")

```

```python
def update_grid_tc_csv(lon, lat, tc_sid, max_wind_speed, entry_time, exit_time, tc_name):
    grid_file = os.path.join(output_dir, f"grid_{lon:.16f}_{lat:.16f}.csv")
    data = pd.DataFrame({
        'tc_sid': [tc_sid],
        'max_wind_speed': [max_wind_speed],
        'start_time': [str(entry_time)],
        'end_time': [str(exit_time)],
        'SID': [tc_sid],  # Duplicate of tc_sid as per request
        'Name': [tc_name]
    })

    # Append to existing CSV or create new one
    if os.path.exists(grid_file):
        existing_df = pd.read_csv(grid_file)
        updated_df = pd.concat([existing_df, data], ignore_index=True)
        updated_df.to_csv(grid_file, index=False, float_format='%.16f')
    else:
        data.to_csv(grid_file, index=False, float_format='%.16f')
```

```python
# Function to process a single NetCDF file
def process_tc_file(file_path, buffer_geometry, gdf_points, gdf_land, gdf_lines):
    # Extract TC_SID from filename
    tc_sid = Path(file_path).stem.replace("ERA5_Land_", "")

        # Get entry and exit times
    entry_time, exit_time, tc_name = get_entry_exit_times(tc_sid, gdf_points, buffer_geometry)
    if entry_time is None or exit_time is None:
        print(f"No buffer entry/exit found for {tc_sid}, skipping.")
        return None, tc_sid, None, None, None

    # Load NetCDF file
    ds = xr.open_dataset(file_path)

    # Filter time range
    ds = ds.sel(valid_time=slice(entry_time, exit_time))
    if len(ds['valid_time']) == 0:
        print(f"No data in time range for {tc_sid}, skipping.")
        ds.close()
        return None, tc_sid, None, None, None

    # Calculate wind speed (m/s) from u10 and v10
    wind_speed = np.sqrt(ds['u10']**2 + ds['v10']**2)

    # Get coordinates
    lats = ds['latitude'].values
    lons = ds['longitude'].values
    times = ds['valid_time'].values

    # Create mask for points within buffer
    mask = np.zeros((len(lats), len(lons)), dtype=bool)
    for i, lat in enumerate(lats):
        for j, lon in enumerate(lons):
            mask[i, j] = is_within_buffer(lon, lat, buffer_geometry)

    # Apply mask to wind speed (set points outside buffer to NaN)
    wind_speed_masked = wind_speed.where(xr.DataArray(mask, coords={'latitude': lats, 'longitude': lons}, dims=['latitude', 'longitude']))

    # Compute maximum wind speed per grid cell over time
    max_wind_speed = wind_speed_masked.max(dim='valid_time', skipna=True)

    # Convert to wind gust by multiplying by 1.6
    max_wind_gust = max_wind_speed * 1.6


    # Plot wind footprint
    plot_wind_footprint(max_wind_gust, tc_sid, entry_time, exit_time, gdf_land, gdf_lines)


     # Convert to DataFrame 
    lons_grid, lats_grid = np.meshgrid(lons, lats)
    df = pd.DataFrame({
        'latitude': lats_grid.flatten(),
        'longitude': lons_grid.flatten(),
        'max_wind_gust': max_wind_gust.values.flatten(),
        'tc_sid': tc_sid,
        'start_time': str(entry_time),
        'end_time': str(exit_time),
        'name': tc_name
    })

    # Format coordinates to 16 decimal places
    df['latitude'] = df['latitude'].map('{:.16f}'.format)
    df['longitude'] = df['longitude'].map('{:.16f}'.format)

    # Update per-grid CSVs
    for _, row in df.dropna().iterrows():
        update_grid_tc_csv(float(row['longitude']), float(row['latitude']), tc_sid, row['max_wind_gust'], entry_time, exit_time, tc_name)

    # Save TC event max wind gust to CSV
    output_file = os.path.join(output_dir, f"max_wind_gust_{tc_sid}.csv")  
    df[['latitude', 'longitude', 'max_wind_gust']].dropna().to_csv(output_file, index=False, float_format='%.16f')  
    print(f"Saved max wind gust for {tc_sid} to {output_file}")

    # Close dataset
    ds.close()

    return df, tc_sid, entry_time, exit_time, tc_name
```

```python
# Function to aggregate maximum wind speeds across all TC events
def aggregate_max_wind_speeds(tc_dfs):
    if not tc_dfs:
        print("No valid TC data to aggregate.")
        return

    # Combine all DataFrames
    combined_df = pd.concat(tc_dfs)

    # Convert lat/lon back to float for grouping
    combined_df['latitude'] = combined_df['latitude'].astype(float)
    combined_df['longitude'] = combined_df['longitude'].astype(float)

    # Group by lat/lon to find max wind speed and corresponding TC metadata
    max_wind_df = combined_df.groupby(['latitude', 'longitude']).apply(
        lambda x: pd.Series({
            'max_wind_speed': x['max_wind_speed'].max(),
            'tc_sid': x.loc[x['max_wind_speed'].idxmax(), 'tc_sid'],
            'start_time': x.loc[x['max_wind_speed'].idxmax(), 'start_time'],
            'end_time': x.loc[x['max_wind_speed'].idxmax(), 'end_time'],
            'name': x.loc[x['max_wind_speed'].idxmax(), 'name']
        })
    ).reset_index()

    # Format coordinates to 16 decimal places
    max_wind_df['latitude'] = max_wind_df['latitude'].map('{:.16f}'.format)
    max_wind_df['longitude'] = max_wind_df['longitude'].map('{:.16f}'.format)

    # Save each grid cell to its own CSV
    for _, row in max_wind_df.iterrows():
        grid_file = os.path.join(output_dir, f"{row['longitude']}_{row['latitude']}.csv")
        data = pd.DataFrame([{
            'latitude': row['latitude'],
            'longitude': row['longitude'],
            'max_wind_speed': row['max_wind_speed'],
            'start_time': row['start_time'],
            'end_time': row['end_time'],
            'SID': row['tc_sid'],
            'Name': row['name']
        }])
        data.to_csv(grid_file, index=False, float_format='%.16f')
        print(f"Saved aggregated max wind speed to {grid_file}")

```
